This dataset compiles stats from pokemon across 7 generations of games. The traget feature I chose was the base total for the stats and the relationship between them. Below shows all columns and what each column contains inside this dataset, 

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import statsmodels.api as sm


df = pd.read_csv('pokemon.csv')
df.info()
df.head()
```

This dataset compiles stats from pokemon across 7 generations of games. The traget feature I chose was the base total for the stats and the relationship between them. 


```python
sns.regplot(x='attack',y='base_total',data=df)
plt.show()
x = sm.add_constant(df['attack'])
y = df['base_total']

model = sm.OLS(y,x).fit()

model.summary()
```

```python
sns.regplot(x='defense',y='base_total',data=df)
plt.show()
x = sm.add_constant(df['defense'])
y = df['base_total']

model = sm.OLS(y,x).fit()

model.summary()
```
